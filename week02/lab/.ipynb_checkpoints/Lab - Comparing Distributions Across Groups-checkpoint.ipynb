## Step 3: Operation


### Make the borough names human-readable

The source stores borough names in all capital letters. **Title case** capitalizes the first letter of each word and writes the remaining letters in lowercase. For example, `STATEN ISLAND` becomes `Staten Island`.

Use `.str.title()` to convert every value in `requests["borough"]` to title case and save the result back into the same feature.


```python
requests["borough"] = requests["borough"].str.title()
```

```python
requests.head()
```

**What roles do `.str` and `.title()` play in this statement?**

Your answer:.str uses Pandas' string methods for a series, treating each value like text.
.title() is a specific string method that converts each name to a title case


### Create a calendar date

Create `requests["date"]` with `requests["created_date"].dt.normalize()`. The `.dt` accessor opens Pandas' datetime tools, and `.normalize()` sets the time in every value to midnight while preserving its calendar date.


```python
requests["date"] = requests["created_date"].dt.normalize()
```

```python
requests.head()
```

### Count requests for each borough and date

Create a DataFrame named `daily_requests` with this sequence:

1. group `requests` by `date` and `borough`;
2. use `.size()` to count rows in every group; and
3. use `.reset_index(name="request_count")` to return a DataFrame whose count column is named `request_count`.

Display the first five rows.


```python
daily_requests = (
    requests
    .groupby(["date", "borough"])
    .size()
    .reset_index(name="request_count")
)

daily_requests.head()
```

**What does one row of `daily_requests` represent? How is that different from one row of `requests`?**

Your answer:It represents the total number of street condition requests for a specific borough on a specific calender date.


**Why did we remove the time before grouping and counting the requests?** As a hint, try repeating the calculation with `created_date` in place of `date` and inspect the result.

Your answer:Because it would group by every unique timestamp rather than just by the day


### Set the borough order

Create `borough_order` containing the five borough names in alphabetical order.


```python
borough_order = sorted(daily_requests["borough"].unique())
borough_order
```

### Calculate the mean daily request count

Create `borough_summary` by grouping `daily_requests` by `borough`, selecting `request_count`, calculating the mean, and using `.reindex(borough_order)`.


```python
borough_summary = (
    daily_requests
    .groupby("borough")["request_count"]
    .mean()
    .reindex(borough_order)
)

borough_summary
```

**Why is comparing only these means incomplete?**

Your answer:Because two boroughs can have similar mean daily counts but very different spreads, skewness, or frequency of unusually high or low days.

Count the number of rows for each borough in `daily_requests`, then use `.reindex(borough_order)`.


```python
daily_counts_per_borough = (
    daily_requests["borough"]
    .value_counts()
    .reindex(borough_order)
)

daily_counts_per_borough
```

### Check the calculated data

Display missing-value counts for `date`, `borough`, and `request_count`. Then use `.describe()` to summarize `request_count` within each borough and put the boroughs in `borough_order`.


```python
daily_requests[["date", "borough", "request_count"]].isna().sum()

daily_requests_summary = (
    daily_requests
    .groupby("borough")["request_count"]
    .describe()
    .reindex(borough_order)
)

daily_requests_summary
```

## Step 5: Evidence


### Create overlapping histograms

Use `sns.histplot()` with:

- `data=daily_requests`;
- `x="request_count"`;
- `hue="borough"` and `hue_order=borough_order`;
- `element="step"`;
- layered histograms;
- `alpha=0.45`; and
- the `"deep"` palette.

Capitalize the legend title and add a question-focused title and descriptive axis labels. Label the y-axis `Number of days`.


```python
plt.figure(figsize=(10, 6))
sns.histplot(
    data=daily_requests,
    x="request_count",
    hue="borough",
    hue_order=borough_order,
    element="step",
    alpha=0.45,
    palette="deep"
)

plt.title("Daily Street-Condition Request Counts by Borough, NYC 2025")
plt.xlabel("Daily number of street-condition requests")
plt.ylabel("Number of days")
plt.tight_layout()
plt.show()
```

**What does the histogram reveal, and what makes the boroughs easy or difficult to compare?**

Your answer:The histogram shows the overall distribution of daily request counts for each borough—how often low, medium, and high daily counts occurred throughout 2025. It reveals which boroughs tend to have higher typical daily counts, how wide their ranges are, and where their distributions overlap.

The boroughs are easy to compare because the histogram places all five distributions on the same axis, letting you see general patterns like which borough usually has more requests and which has more variability.


### Create a box plot

Put `request_count` on the x-axis and `borough` on the y-axis. Use `borough_order`, the named color `"mediumseagreen"`, and descriptive labels.


```python
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=daily_requests,
    x="request_count",
    y="borough",
    order=borough_order,
    color="mediumseagreen"
)

plt.title("Daily Street-Condition Requests by Borough, NYC 2025")
plt.xlabel("Daily number of street-condition requests")
plt.ylabel("Borough")
plt.tight_layout()
plt.show()
```

**What differences among boroughs are easier to see in the box plot than in the histogram?**

Your answer: The box plot makes it easier to compare medians, interquartile ranges, and outliers across the boroughs.


### Create a violin plot

Use the same x feature, y feature, and order. Change only the plot color to `"mediumpurple"` in addition to the required data mappings and order.


```python
plt.figure(figsize=(10, 6))
sns.violinplot(
    data=daily_requests,
    x="request_count",
    y="borough",
    order=borough_order,
    color="mediumpurple"
)

plt.title("Shape of Daily Street-Condition Request Distributions by Borough, NYC 2025")
plt.xlabel("Daily number of street-condition requests")
plt.ylabel("Borough")
plt.tight_layout()
plt.show()
```

**What does the violin plot add to the comparison?**

Your answer: It shows the full distibution shape of each burrow, including density, skewness, and whether the distribution has multiple peaks.


**Which of the three figures most clearly compares daily request counts among boroughs? Explain your choice.**

Your answer The box plot provides the clearest comparison.

# Your own analysis

Develop a new research question using the 311 data. Your question must compare a **quantitative feature across categorical groups**. You will probably need to calculate the quantitative feature from the original request records.

Possible grouping features include `descriptor`, month, or day of the week. You may use another meaningful categorical feature that you can create and explain.


## Step 1: Question


**Write a research question that compares a quantitative feature across categorical groups.**

Your answer: How do daily street-condition request counts differ across days of the week in 2025?


## Step 2: Data


**Which original features will you use, and what will each row of your calculated DataFrame represent?**

Your answer: I will use created_date to derive the weekday and request_id to count requests. Each row of my calculated DataFrame will represent the number of street-condition requests for a specific date and weekday.


## Step 3: Operation

Prepare the grouping feature, calculate the quantitative feature, and save the resulting DataFrame with a meaningful name.


```python
# Create weekday feature
requests["weekday"] = requests["created_date"].dt.day_name()

# Count daily requests by weekday
daily_weekday_requests = (
    requests
    .groupby(["date", "weekday"])
    .size()
    .reset_index(name="request_count")
)

# Order weekdays
weekday_order = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

# Summarize mean daily counts
weekday_summary = (
    daily_weekday_requests
    .groupby("weekday")["request_count"]
    .mean()
    .reindex(weekday_order)
)

weekday_summary
```

**Briefly explain how your code produced the quantitative feature and comparison groups.**

Your answer: The code creates a weekday category using .dt.day_names(), grouped by date and weekday count requests. Then calculates the mean daily request count for each weekday.


## Step 4: Check

Check the calculated data for missing or impossible values, confirm that the intended groups are present, and use an appropriate numerical summary to inspect the distributions.


```python
daily_weekday_requests.isna().sum()

daily_weekday_requests["weekday"].value_counts()

weekday_summary.describe()
```

**What did your checks establish?**

Your answer: it confirms there is no missing values, all seven weekdays are present.


## Step 5: Evidence

Create a histogram, box plot, or violin plot that directly addresses your question. Include a clear title and descriptive axis labels.


```python
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=daily_weekday_requests,
    x="weekday",
    y="request_count",
    order=weekday_order,
    color="steelblue"
)

plt.title("Daily Street-Condition Requests by Day of Week, NYC 2025")
plt.xlabel("Day of week")
plt.ylabel("Daily number of street-condition requests")
plt.tight_layout()
plt.show()
```

**What pattern in the figure provides the strongest evidence for your answer?**

Your answer: The strongest evidence comes from the differences in medians and spreads across weekdays. If weekdays show higher medians than weekends, that indicates more requests are submitted during the workweek.


## Step 6: Conclusion


**State one conclusion that directly answers your research question.**

Your answer: Daily street-condition request counts are higher and more variable on weekdays than weekends, suggesting that reporting activity peaks during the workweek.


## Step 7: Limitation


**State one important limitation on how your result should be interpreted.**

Your answer: This reflects reporting patterns, not necessarily when street problems occur.
